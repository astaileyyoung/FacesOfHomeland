```python
import ast 

import dlib
import numpy as np
import pandas as pd
import networkx as nx
from tqdm import tqdm 
from chinese_whispers import aggregate_clusters, chinese_whispers
```

```python
G = nx.Graph()
```

```python
df = pd.read_csv('./data/faces.csv', index_col=0)
df.head()
```

```python
episode = df[(df['season'] == 1) &
             (df['episode'] == 1)]
vectors = [dlib.vector(np.array(ast.literal_eval(x))) for x in tqdm(episode['encoding'].tolist())]
```

```python
G.add_nodes_from(vectors)
```

```python
G = nx.karate_club_graph()
list(G.edges)
```

```python
chinese_whispers(G, weighting='top')
```

```python
for label, cluster in sorted(aggregate_clusters(G).items(), key=lambda e: len(e[1]), reverse=True):
    print(f"{label}\t{cluster}\n")
```

```python
%config InlineBackend.figure_formats = ['svg']

import matplotlib.pyplot as plt
```

```python
colors = [1. / G.nodes[node]["label"] for node in G.nodes()]

nx.draw_networkx(G, cmap=plt.get_cmap("jet"), node_color=colors, font_color="white")
```
